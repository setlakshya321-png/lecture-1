```python
#top 10 profit/loss product
df["Total_Cost_Lakh"]=(
    df["Marketing_Spend_Lakh"]
    +df["Content_Cost_Lakh"]
    +df["Salary_Cost_Lakh"]
    +df["Other_Cost_Lakh"]
)
df["Profit_Lakh"] = df["Revenue_Lakh"] - df["Total_Cost_Lakh"]
print (df[["Product", "Revenue_Lakh", "Total_Cost_Lakh", "Profit_Lakh"]].head(10))
```

```python
#Missing Value
print(df.isna().sum())
```

```python
# decision making structures
#IF statements
rating = df ["Customer_Rating"].mean()

if rating >= 4:
  print("Customer satisfaction is good")
```

```python
# decision making structures
#IF statements
rating = df ["Revenue_Lakh"].mean()

if rating > 200:
  print("Revenue is sexy")
```

```python
# decision making structures
#IF statements
rating = df ["Revenue_Lakh"].mean()

if rating >300:
  print("Average revenue is high")
else:
  print("Average revenue is low")
```
